## 1. Hierarquia de necessidades em Analytics

Inspirada na pirâmide de Maslow, mostra que cada etapa depende da anterior. Você só sobe de nível quando a base está sólida.

```
                ┌─────────────────────────────┐
                │  5. Otimizar e prever        │
                │     com Machine Learning     │
                └──────────────┬──────────────┘
             ┌─────────────────┴─────────────────┐
             │  4. Analisar KPIs e gerar insights  │
             └─────────────────┬─────────────────┘
          ┌────────────────────┴────────────────────┐
          │  3. Definir KPIs e acompanhar evolução  │
          └────────────────────┬────────────────────┘
       ┌───────────────────────┴───────────────────────┐
       │              2. Limpar os dados               │
       └───────────────────────┬───────────────────────┘
    ┌──────────────────────────┴──────────────────────────┐
    │                1. Coletar os dados                  │
    └─────────────────────────────────────────────────────┘
```

| Nível | Descrição |
|---|---|
| **1. Coletar** | Fontes, pipelines, ingestão repetível |
| **2. Limpar** | Nulos, duplicatas, outliers, inconsistências |
| **3. KPIs** | Métricas mensuráveis ligadas ao negócio |
| **4. Insights** | Explorar *por quê* algo mudou |
| **5. ML** | Prever, recomendar, automatizar decisões |

> **Ideia central:** pular etapas (ir direto para ML sem limpeza ou KPIs claros) costuma gerar projetos caros que não sustentam decisões no dia a dia.

## 2. Quando usar Machine Learning?

**ML existe para economizar tempo** — automatizar ou acelerar trabalho que seria lento, repetitivo ou caro. Não é um fim em si; precisa gerar **valor de negócio** mensurável.

### Escada de complexidade

```
  Regras simples (heurísticas simples)
           ↓
  Regras mais elaboradas (heurísticas complexas)
           ↓
  Machine Learning
```

Comece pelo mais simples que resolve o problema. **Só suba de degrau quando o anterior não for suficiente.**

| Abordagem | Quando considerar |
|---|---|
| **Heurísticas simples** | Poucos dados, regras óbvias ("se X, então Y") |
| **Heurísticas complexas** | Muitas condições, mas ainda explicáveis em linguagem de negócio |
| **Machine Learning** | Muito dado, muitas variáveis, padrões difíceis de codificar à mão |

### Antes de escolher ML, verifique

1. **Tamanho dos dados** — volume suficiente para generalizar?
2. **Variáveis explicativas (sinais)** — features disponíveis, confiáveis, sem vazamento de futuro?
3. **Poder explicativo** — os sinais realmente se relacionam com o alvo?
4. **Valor de negócio** — qual decisão melhora? como medir sucesso?
5. **Baseline sem ML** — solução simples já atende? ML pode ser overkill.

### A regra dos 80/20

```
  Solução simples  ──────────►  ~80% do valor desejado
         │
         │  (muito esforço extra)
         ▼
  Modelo de ML     ──────────►  dos 80% aos 100% (ganhos marginais)
```

> Pergunta honesta: *os últimos 20% justificam semanas/meses de engenharia, infra e manutenção?*

### Sinais de que ML faz sentido

Use ML quando **vários** destes pontos forem verdadeiros:

- Muito dado (volume/variedade que humanos não processam bem)
- Problema repetitivo que consome muito tempo
- Caro resolver só com pessoas (escala, especialistas escassos)
- Heurísticas já não escalam — regras ficaram frágeis
- Há retorno claro (receita, custo, risco, UX)

> **Resumo:** ML para ganhar tempo, não para impressionar. Teste o simples antes.

## 3. Exemplo: otimizar recomendações para subir retenção

Cenário: *"Podemos aumentar a **retenção** se otimizarmos o mecanismo de **recomendação**."*

### Passo 1 — Definir o problema com clareza (crítico)

- O que é **retenção** aqui? (D7? D30? assinatura ativa?)
- O que significa **otimizar recomendação**? (cliques, tempo na sessão, compra?)
- Qual hipótese liga recomendação → retenção?
- O que **não** é escopo?

> Definição incorreta → equipe treina o modelo certo para a pergunta errada.

### Passo 2 — Correlação e features

- Análise de correlação entre sinais e o alvo
- Listar o que o usuário fez, viu, comprou, quando, em qual dispositivo
- Priorizar features com poder explicativo disponíveis em produção

### Passo 3 — Plano de projeto

| Item | O que definir |
|---|---|
| Intervalo de dados | Treino / validação / teste temporal |
| Volume | Usuários, eventos, itens |
| Features | Lista, origem, missing |
| Modelos | Baselines + candidatos |
| Produção | Batch vs. tempo real, SLA, fallback |
| Critério de vencedor | Métrica offline + KPI de negócio |

### Passo 4 — Ciclo de experimentação

```
  Explorar → treinar candidatos → comparar métricas → iterar
                      ↑________________________________|
```

### Passo 5 — Produção

1. **API REST** que devolve predições rápido o suficiente
2. **Teste online** (shadow mode, A/B, canary)
3. **Lançamento** gradual com rollback planejado

> Hipótese → problema definido → features → plano → experimentos → API → testes → **launch**.

## 4. Supervisionado vs. Não supervisionado

| | **Supervisionado** | **Não supervisionado** |
|---|---|---|
| Objetivo | Prever um resultado específico | Descobrir estrutura nos dados |
| Gabarito | ✅ Humanos fornecem rótulos | ❌ Algoritmo cria os grupos |
| Analogia | Estudar com caderno de respostas | Agrupar sem saber as categorias antes |

### Supervisionado

O modelo aprende a relação entre **features (entradas)** e **rótulo (saída)** usando exemplos rotulados.

**Exemplo — classificar mamão:**
- Features: maciez, comprimento, massa, teor de açúcar
- Rótulo: "é mamão" / "não é mamão"
- Em produção: o modelo classifica frutas novas só pelas medições

**Outros casos:** churn (sim/não), preço de imóvel (regressão), spam (spam/não spam)

### Não supervisionado

Sem rótulo — o algoritmo descobre padrões, grupos ou representações por conta própria.

**Exemplo — segmentos de audiência:**
- Entrada: comportamento (idade, interesses, renda) sem rótulo de segmento
- Output: grupos como "homens jovens, amam esportes", "público familiar", etc.
- Esses clusters **não foram definidos manualmente**

| Pergunta | Supervisionado | Não supervisionado |
|---|---|---|
| Temos a resposta no treino? | **Sim** | **Não** |
| O que o modelo faz? | **Prevê** o rótulo | **Descobre** grupos |
| Risco principal | Rótulos errados/desbalanceados | Clusters difíceis de interpretar |

## 5. Regressão vs. Classificação

Ambos são **aprendizado supervisionado**. A diferença está no **tipo de variável alvo**.

| | **Regressão** | **Classificação** |
|---|---|---|
| Alvo | Variável **contínua** | Variável **categórica** |
| Pergunta | *"Quanto?"* | *"Qual categoria?"* |
| Saída | Número (ex: R$ 1.240) | Classe ou probabilidade (ex: "churn", 78%) |

### Regressão — exemplos reais

| Domínio | O que prever |
|---|---|
| **Uber** | ETA (tempo estimado de chegada) |
| **Airbnb** | Preço sugerido para o imóvel |
| **Ads** | CPM, CPC, conversão em R$ |
| **Negócio** | Customer Lifetime Value (CLV) |
| **Vendas** | Receita, volume de vendas |

**Métricas:** MAE, RMSE, R²

### Classificação — exemplos reais

| Domínio | O que classificar |
|---|---|
| **Retenção** | Usuário vai churnar? (sim/não) |
| **Conversão** | Visitante comprou? |
| **E-mail** | E-mail foi aberto? |
| **Sentimento** | Feedback: positivo / negativo / neutro |

**Métricas:** Acurácia, Precisão, Recall, F1, AUC-ROC

### Product Analytics

```
  Product analytics
        │
        ├── Regressão     → KPIs numéricos (CLV, revenue, ETA…)
        └── Classificação → KPIs categóricos (retido?, converteu?, sentimento…)
```

> **Regressão** = prever **quanto**. **Classificação** = prever **qual categoria**.

## 6. Product Analytics — fluxo de trabalho

```
  Pré-processamento → EDA → Feature engineering → Modelagem → Hipóteses
        ↑________________________________________________________________|
                              (iterar)
```

| Etapa | Pergunta que responde |
|---|---|
| **Pré-processamento** | Os dados estão limpos e confiáveis? |
| **EDA** | O que está acontecendo e onde olhar? |
| **Feature engineering** | Quais sinais descrevem o comportamento? |
| **Modelagem** | O que conseguimos prever ou segmentar? |
| **Hipóteses / insights** | O que o time de produto deve **testar** ou **mudar**? |

### Criar hipóteses que podemos rodar

**Formato útil:**
> *Se [intervenção], então [métrica] vai [subir/cair] em [segmento] em [prazo], porque [razão baseada em dados].*

**Checklist antes de rodar:**
- [ ] Métrica primária definida (ex.: retenção D7)
- [ ] Segmento claro (novos usuários, mobile, plano free)
- [ ] Tamanho de amostra razoável
- [ ] Duração do teste definida
- [ ] Tracking confiável

**Exemplos:**
1. *Notificação no dia 1 aumenta retorno D3 em usuários sem tutorial completo.*
2. *Recomendações por cluster aumentam cliques vs. lista genérica.*

## 7. ML para Data Products

Um **produto de dados** usa ML para entregar valor contínuo: recomendações, previsões, scores de risco, busca inteligente.

```
  Enquadramento → Coleta → Processamento → EDA → Features
       → Validação cruzada → Modelo → Avaliação → Produtização
```

### Tensão central: Acurácia vs. Interpretabilidade

| Prioridade | Quando | Abordagem |
|---|---|---|
| **Interpretabilidade** | Regulação, debug, stakeholders não técnicos | Regressão logística, árvores, SHAP |
| **Acurácia** | Escala, personalização fina | Ensembles, deep learning, embeddings |
| **Equilíbrio** | Maioria dos casos de produto | Modelo complexo + relatório de features importantes |

### Tipos comuns

| Tipo | Exemplos | Métricas |
|---|---|---|
| **Recommender** | Vídeos, produtos, rotas | Clique, tempo, conversão |
| **Forecasting** | Demanda, receita, ETA | MAE, RMSE, MAPE |
| **Score / decisão** | Churn, fraude, lead score | AUC, precisão, recall |

### Checklist — ML em Data Product
- [ ] Problema e KPI definidos
- [ ] Dados coletados e pipeline estável
- [ ] EDA e features validadas (sem leakage)
- [ ] Cross-validation adequada ao tipo de dado
- [ ] Modelo avaliado offline **e** plano de teste no produto
- [ ] Produtizado: serve predição, monitora, retreina

> **Data product** = modelo + dados + engenharia + métrica de negócio.
